### 1. Find prime numbers from a list of numbers

```python
numbers = range(0, 15)
prime_numbers = list()

for num in numbers:
    count = 0
    for devider in range(2, num):
        if num % devider == 0:
            count += 1
            break
    if count == 0:
        prime_numbers.append(num)

print(f"The all prime numbers are : {prime_numbers}")
```

### Way - 2

```python
numbers = range(0, 15)
prime_numbers = set(numbers)

for num in numbers:
    for devider in range(2, num):
        if num % devider == 0:
            prime_numbers.remove(num)
            break
print(f"The all prime numbers are : {prime_numbers}")
```

```python
prime_numbers = [num for num in numbers if not all([True if num % d == 0 else False for d in range(2, num)])]
print(f"The all prime numbers are : {prime_numbers}")
```
